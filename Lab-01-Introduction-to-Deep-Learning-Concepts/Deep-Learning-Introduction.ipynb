## 1. What is Deep Learning?

Deep Learning is a subset of Machine Learning that uses artificial neural networks with multiple layers.

These networks learn patterns from data. Instead of manually selecting every feature, a deep learning model can learn useful features during training.

### Relationship

Artificial Intelligence (AI) contains Machine Learning (ML), and Machine Learning contains Deep Learning (DL).

### Common applications
- Image and face recognition
- Speech recognition
- Natural language processing
- Medical image analysis
- Self-driving vehicles


## 2. Evolution of Deep Learning

| Period | Development |
|---|---|
| 1950s–1980s | Early neural networks and the Perceptron |
| 1980s | Backpropagation gained wider attention |
| 1990s | Traditional ML methods became widely used |
| 2000s | More data and computing power supported neural network research |
| 2012 | AlexNet achieved a major ImageNet breakthrough |
| 2014 onward | Deep learning research and applications expanded rapidly |

Important contributors include Frank Rosenblatt, Geoffrey Hinton, Yann LeCun, Andrew Ng, and Ian Goodfellow.


## 3. Artificial Neural Network (ANN)

An Artificial Neural Network is inspired by the way biological neurons communicate.

### Main layers
1. **Input layer:** receives the input data.
2. **Hidden layer:** processes information and learns patterns.
3. **Output layer:** produces the final prediction.

### Important terms
- **Neuron:** a processing unit in a network.
- **Weight:** a value that controls the strength of a connection.
- **Bias:** an additional adjustable value.
- **Activation function:** introduces non-linearity.
- **Epoch:** one complete pass through the training data.
- **Learning rate:** controls the size of weight updates.
- **Backpropagation:** calculates gradients to help update network parameters.


## 4. Important Deep Learning Milestones

## 5. Traditional Machine Learning vs Deep Learning

| Traditional Machine Learning | Deep Learning |
|---|---|
| Often depends on manually designed features | Can learn features automatically |
| Can work with smaller datasets | Often benefits from large datasets |
| Usually requires fewer computing resources | Often requires more computing resources |
| Examples: Logistic Regression, Decision Trees | Examples: CNNs, RNNs, Transformers |

### Sentiment analysis example

Traditional ML can convert text into numerical features using Bag of Words and then train a classifier.

Deep Learning can learn richer representations of words and sentences using neural networks.
